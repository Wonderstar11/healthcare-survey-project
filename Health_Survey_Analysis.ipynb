```python
gender_counts = df_clean["gender"].value_counts()

print(gender_counts)
```

```python
spending_by_gender = df_clean.groupby("gender")[
    ["utilities", "entertainment", "school_fees", "shopping", "healthcare"]
].mean()

display(spending_by_gender)
```

```python
spending_by_gender.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Average Spending by Gender Across Expense Categories")
plt.xlabel("Gender")
plt.ylabel("Average Expenditure (₦)")
plt.xticks(rotation=0)
plt.legend(title="Expense Categories")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()

plt.savefig("gender_spending_comparison.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(9, 6))

plt.scatter(
    df_clean["age"],
    df_clean["total_income"],
    s=100
)

plt.title("Relationship Between Age and Total Income")
plt.xlabel("Age (Years)")
plt.ylabel("Total Income (₦)")
plt.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("age_income_relationship.png", dpi=300)
plt.show()
```

## Healthcare Survey Data Analysis Summary

### 1. Dataset Overview

The survey dataset initially contained nine records and eight columns. During data cleaning, four duplicate records were identified and removed, leaving five unique responses for analysis.

The dataset contains respondents' ages, gender, total income and expenditure across five categories: utilities, entertainment, school fees, shopping and healthcare.

### 2. Descriptive Analysis

The average age of respondents was 37.2 years, while the average monthly income was ₦332,000.

School fees recorded the highest average expenditure at ₦89,000, followed by shopping at ₦62,000, utilities at ₦37,000, healthcare at ₦36,400 and entertainment at ₦24,000.

### 3. Gender Distribution

The dataset contained two male respondents (40%), two female respondents (40%) and one respondent who identified as Other (20%).

### 4. Spending Patterns by Gender

The analysis showed that male respondents had higher average expenditure than female respondents across all five spending categories.

Male respondents recorded average expenditure of ₦42,500 on utilities, ₦27,500 on entertainment, ₦112,500 on school fees, ₦42,500 on shopping and ₦22,500 on healthcare.

Female respondents recorded average expenditure of ₦17,500 on utilities, ₦10,000 on entertainment, ₦35,000 on school fees, ₦12,500 on shopping and ₦18,500 on healthcare.

The Other category contained only one respondent, whose spending figures should therefore be interpreted as individual observations rather than a group average.

### 5. Relationship Between Age and Income

The age-income visualisation showed that the 56-year-old respondent had the highest recorded income of ₦560,000, followed by the 42-year-old respondent with ₦450,000.

The 35-year-old respondent earned ₦300,000, the 32-year-old earned ₦250,000, while the 21-year-old respondent had the lowest recorded income of ₦100,000.

The observations suggest a possible positive relationship between age and income within this small dataset, although a larger sample would be needed to establish this reliably.

### 6. Conclusion and Limitation

The analysis demonstrates how survey data can be collected through a Flask application, stored in MongoDB, exported to CSV, cleaned and analysed using Python.

However, the dataset contains only five unique responses. Consequently, the findings are preliminary and cannot be used to make reliable generalisations about the spending habits of a wider population. A larger and more representative sample would be required for stronger conclusions.
